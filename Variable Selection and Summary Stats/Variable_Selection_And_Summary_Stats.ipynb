### Selecting Columns of Interest

```python
cols = [
    #who/when
    "PlayerName", "position", "Throws", "Season", "Age",
    #workload
    "IP", "GS",
    #results
    "ERA", "FIP", "WHIP", "SO", "K%", "BB%",
    #value
    "WAR", "Dollars",
    #fastball velocity
    "FBv",
    #contact quality allowed
    "HardHit%", "Barrel%",
]

missing = [c for c in cols if c not in df.columns]
print("Not found:", missing)

variables = df[cols].copy()
print(variables.shape)
variables.head()
```

### Cleaning the Data

Filtering

```python
#This data looks at anyone who has thrown a pitch. Due to this, the data needs to be filtered to only true pitchers. 
#This will get rid of position players who came in and threw for 1 inning
variables = variables[variables.position == 'P']
```

Checking summary stats for games started and innings pitched

```python
variables.GS.describe()
```

```python
variables.IP.describe()
```

```python
#Given that Tarik Skubal is a starter, we want to make sure we are only looking at starters and not relievers
#as they receive vastly different contracts. In addition, we want to make sure we are getting close to a season's worth
#of pitching for a starter as opposed to having seasons in which a starter only made 8 starts and then got hurt. 
#Any row that was from the 2020 season will also be dropped from our data for a similar reason as there were only 60 games
#that year and each pitcher made at most a handful of starts.
variables = variables[
    (variables["position"] == "P")
    & (variables["Season"] != 2020)
    & (variables["GS"] >= 25)
    & (variables["IP"] >= 130)
].copy()

print(variables.shape)
#We have roughly an equal amount of pitching rows from each year
print(variables["Season"].value_counts().sort_index())
```

```python
#Now that the data is starters who pitched roughly a full season, we can begin to inspect the data for any NAs
variables.isna().sum().sort_values(ascending=False)
```

```python
#There are no NAs for any of the selected columns, so the data is now clean. 
```

### Creating Summary Stats Table

```python
num_cols = ["Age", "IP", "GS", "ERA", "FIP", "WHIP", "SO", "K%", "BB%",
            "WAR", "Dollars", "FBv", "HardHit%", "Barrel%"]

summary = variables[num_cols].describe().T[["count", "mean", "std", "min", "50%", "max"]]
summary = summary.rename(columns={"50%": "median"}).round(2)
summary
```

```python
#Saving data as a CSV
variables.to_csv("variables.csv", index=False)
```
